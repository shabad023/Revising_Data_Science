# Bias Variance Trade off

Variance is just how scattered the plot is,

Bias is ke bullseye se kitna door hai 

```python
# High Bias is UnderFitting
# You always shoot to almost the same place, but it’s far from the center.
# The arrows are close together, but all are wrong.

# Meaning:

#  The model is too simple.
#  It doesn’t learn enough from the data.
#  It makes similar mistakes every time.
```

```python
# High Variance (Overfitting)
# Now imagine every arrow lands in a completely different place
# Sometimes close.

# Sometimes very far.

# This is High Variance.

# Meaning:

# * The model learns the training data too well.
# * It even memorizes noise.
# * It performs poorly on new data.

# Example:

# Instead of learning the trend, it remembers every point.
# A very wiggly curve that passes through almost every point.
```

```python
# Case 3 - A Goood Model-- (Bias-Variance Balance)
# The arrows are:

# * Near the center ✅
# * Close to each other ✅
```

Easy Analogy: Studying for an Exam

High Bias Student

Studies only one page.

Exam has 20 chapters.
Didn’t learn enough.
Training Test  ❌
Real Exam      ❌


High Variance Student

Memorizes every single practice question.

Real exam has different questions.

Result:Memorized instead of understanding.
Practice Test ✅

Real Exam ❌


Balanced Student

Understands the concepts.

Can solve new questions.
Practice Test ✅

Real Exam ✅

```python
def impute_age(cols):
    Age=cols[0]
    Pclass=cols[1]

    if pd.isnull(Age):
        if Pclass==1:
            return 37
        elif Pclass==2:
            return 29
        else:
            return 24

    else:
        return Age
        
```

## Bohot basic jehi gall aa haha

kmeans.cluster_centers_

This stores the coordinates of the centroids of the group .

Cluster_centers_ contains one center for each cluster (group), NOT for each data point.



kmeans.labels_

This tells you

Which cluster each data point belongs to.


So if you have:

* 1000 points and 5 clusters:
    * len(kmeans.labels_) = 1000
    * len(kmeans.cluster_centers_) = 5
